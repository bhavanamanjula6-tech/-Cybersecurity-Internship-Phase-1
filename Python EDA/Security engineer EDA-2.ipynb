```python

import pandas as pd
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

# MySQL connection details
username = "root"
password = "YOUR_MYSQL_PASSWORD"  # Enter your actual MySQL password here
host = "localhost"
port = 3306
database = "cybersecurity_clean"

# Create connection URL safely
connection_url = URL.create(
    drivername="mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(connection_url)

# Test the connection
try:
    with engine.connect() as conn:
        result = conn.execute(
            text("SELECT DATABASE()")
        ).scalar()

        print("Connected successfully!")
        print("Database:", result)

except Exception as e:
    print("Connection failed.")
    print("Error:", e)
```

```python

from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL
from getpass import getpass

username = "root"
password = getpass("Enter your MySQL password: ")

connection_url = URL.create(
    "mysql+pymysql",
    username=username,
    password=password,
    host="localhost",
    port=3306,
    database="cybersecurity_clean"
)

engine = create_engine(connection_url)

try:
    with engine.connect() as conn:
        print("Connection successful!")
        print("Database:", conn.execute(
            text("SELECT DATABASE()")
        ).scalar())

except Exception as e:
    print("Connection failed!")
    print("Error type:", type(e).__name__)
    print("Exact error:", repr(e.orig) if hasattr(e, "orig") else str(e))
```
