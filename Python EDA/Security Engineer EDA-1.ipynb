```python
def clean_text_columns(df, columns):
    df = df.copy()

    for column in columns:
        df[column] = df[column].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )

    return df
```

```python
techniques_clean = techniques_df[
    ["ID", "name", "description", "tactics", "platforms"]
].copy()

techniques_clean = clean_text_columns(
    techniques_clean,
    ["ID", "name", "description", "tactics", "platforms"]
)

print("Techniques cleaned successfully")
```

```python
mitigations_clean = mitigations_df[
    ["ID", "name", "description"]
].copy()

mitigations_clean = clean_text_columns(
    mitigations_clean,
    ["ID", "name", "description"]
)

print("Mitigations cleaned successfully")
```

```python
techniques_addressed_clean = techniques_addressed_df[
    ["source ID", "target ID", "mapping type"]
].copy()

techniques_addressed_clean = clean_text_columns(
    techniques_addressed_clean,
    ["source ID", "target ID", "mapping type"]
)

print("Techniques Addressed cleaned successfully")
```

```python
tactics_clean = tactics_df[
    ["ID", "name"]
].copy()

tactics_clean = clean_text_columns(
    tactics_clean,
    ["ID", "name"]
)

print("Tactics cleaned successfully")
```

```python
from sqlalchemy import text

with engine.connect() as connection:

    print("NULL CHECK IN DATABASE")
    print("----------------------------")

    # techniques
    result = connection.execute(text("""
        SELECT
            SUM(ID IS NULL) AS ID_nulls,
            SUM(name IS NULL) AS name_nulls,
            SUM(description IS NULL) AS description_nulls,
            SUM(tactics IS NULL) AS tactics_nulls,
            SUM(platforms IS NULL) AS platforms_nulls
        FROM techniques
    """))
    print("techniques:", result.fetchone())

    # mitigations
    result = connection.execute(text("""
        SELECT
            SUM(ID IS NULL) AS ID_nulls,
            SUM(name IS NULL) AS name_nulls,
            SUM(description IS NULL) AS description_nulls
        FROM mitigations
    """))
    print("mitigations:", result.fetchone())

    # techniques_addressed
    result = connection.execute(text("""
        SELECT
            SUM(`source ID` IS NULL) AS source_ID_nulls,
            SUM(`target ID` IS NULL) AS target_ID_nulls,
            SUM(`mapping type` IS NULL) AS mapping_type_nulls
        FROM techniques_addressed
    """))
    print("techniques_addressed:", result.fetchone())

    # platforms
    result = connection.execute(text("""
        SELECT
            SUM(name IS NULL) AS name_nulls
        FROM platforms
    """))
    print("platforms:", result.fetchone())

    # tactics
    result = connection.execute(text("""
        SELECT
            SUM(ID IS NULL) AS ID_nulls,
            SUM(name IS NULL) AS name_nulls
        FROM tactics
    """))
    print("tactics:", result.fetchone())
```

```python
from sqlalchemy import text

with engine.connect() as connection:

    print("RELATIONSHIP VALIDATION")
    print("=" * 50)

    # source IDs that don't exist in techniques
    result = connection.execute(text("""
        SELECT COUNT(*)
        FROM techniques_addressed ta
        LEFT JOIN techniques t
            ON ta.`source ID` = t.ID
        WHERE t.ID IS NULL
    """))

    print("Invalid source IDs:", result.scalar())

    # target IDs that don't exist in mitigations
    result = connection.execute(text("""
        SELECT COUNT(*)
        FROM techniques_addressed ta
        LEFT JOIN mitigations m
            ON ta.`target ID` = m.ID
        WHERE m.ID IS NULL
    """))

    print("Invalid target IDs:", result.scalar())
```

```python
from sqlalchemy import text

with engine.begin() as connection:

    # Primary key for techniques
    connection.execute(text("""
        ALTER TABLE techniques
        ADD PRIMARY KEY (ID)
    """))

    # Primary key for mitigations
    connection.execute(text("""
        ALTER TABLE mitigations
        ADD PRIMARY KEY (ID)
    """))

    # Primary key for tactics
    connection.execute(text("""
        ALTER TABLE tactics
        ADD PRIMARY KEY (ID)
    """))

    # Primary key for platforms
    connection.execute(text("""
        ALTER TABLE platforms
        ADD PRIMARY KEY (name)
    """))

    # Composite primary key for techniques_addressed
    connection.execute(text("""
        ALTER TABLE techniques_addressed
        ADD PRIMARY KEY (`source ID`, `target ID`, `mapping type`)
    """))

print("Primary keys added successfully!")
```

```python
from sqlalchemy import text

with engine.begin() as connection:

    # source ID → mitigations.ID
    connection.execute(text("""
        ALTER TABLE techniques_addressed
        ADD CONSTRAINT fk_ta_source_mitigation
        FOREIGN KEY (`source ID`)
        REFERENCES mitigations(ID)
    """))

    # target ID → techniques.ID
    connection.execute(text("""
        ALTER TABLE techniques_addressed
        ADD CONSTRAINT fk_ta_target_technique
        FOREIGN KEY (`target ID`)
        REFERENCES techniques(ID)
    """))

print("Foreign keys added successfully!")
```

```python
from sqlalchemy import text

with engine.connect() as connection:

    print("FINAL KEY VALIDATION")
    print("=" * 60)

    result = connection.execute(text("""
        SELECT
            kcu.TABLE_NAME,
            kcu.COLUMN_NAME,
            kcu.CONSTRAINT_NAME,
            tc.CONSTRAINT_TYPE
        FROM information_schema.KEY_COLUMN_USAGE kcu
        JOIN information_schema.TABLE_CONSTRAINTS tc
            ON kcu.CONSTRAINT_SCHEMA = tc.CONSTRAINT_SCHEMA
            AND kcu.TABLE_NAME = tc.TABLE_NAME
            AND kcu.CONSTRAINT_NAME = tc.CONSTRAINT_NAME
        WHERE kcu.CONSTRAINT_SCHEMA = 'cybersecurity_clean'
        ORDER BY kcu.TABLE_NAME,
                 kcu.CONSTRAINT_NAME,
                 kcu.ORDINAL_POSITION
    """))

    for row in result:
        print(row)
```
